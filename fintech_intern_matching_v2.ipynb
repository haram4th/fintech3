# 인턴십 기업-학생 매칭 결과 정리

본 노트북은 핀테크 과정 수료 학생들의 역량 평가, 선호도 조사, 그리고 기업별 프로젝트 **난이도 및 교육 적합성**을 반영하여 도출한 최종 매칭 결과를 정리한 것입니다.


## 1단계: 기업 프로젝트 난이도·적합도(최종 버전)

학생들이 실제로 학습한 커리큘럼(파이썬, 데이터 수집·분석, LLM·RAG, 기본 웹/기획 이론)을 기준으로 각 기업 프로젝트의 **난이도와 교육 적합성**을 종합 평가한 결과입니다.

| 기업 | 난이도·적합도 점수(100) | 주요 특징 |
|---|---|---|
| **투디지트** | **96** | 미국 상장사·공시 데이터 분석 및 서비스 고도화. 금융 + 데이터 분석 중심, 교육 커리큘럼과 최상위 적합 |
| **씽크플랜두** | **94** | AFG 기반 배달주문 중계 API 기획. 결제 인프라 + API/데이터 흐름, 기술·분석 중심 |

| **더빌딩** | **89** | AI 콜봇·자동화 엔진 기반 임대관리 SaaS. 콜로그·민원 데이터 분석 + 서비스 플로우 설계 |
| **크로스허브** | **83** | SSI 기반 신원인증 및 글로벌 간편결제 서비스. 정통 핀테크 도메인이지만 SSI 개념 등 난이도 중상 |

## 4. 매칭 기본 원칙 정리

1. **기업 난이도와 학생 역량의 적합성 우선**
2. **학생의 1·2·3지망 최대한 존중**
3. **기업당 1~2명 배정 조건 준수**
4. **데이터 분석 및 LLM·API 역량이 살아나는 프로젝트 우선 배정**
5. 난이도 최상(디라브) 및 UX/SSI 중심 고난도 프로젝트(코코넛사일로, 크로스허브)는 교육 목적상 신중하게 배정


## 5. 학생별 최종 매칭 결과 및 근거

### 5-1. 김채민 → **투디지트**
- 역량: 합계 40 (최상위), 파이썬·데이터 분석·모델링 모두 상
- 선호도: 1지망이 투디지트
- 기업 특성: 미국 상장사·공시 데이터 기반 분석 및 서비스 고도화 → 데이터 분석 중심 학생에게 최적
- 결론: **금융 데이터 분석 + 서비스 고도화 경험을 동시에 쌓기 가장 좋은 매칭**

---
### 5-2. 신서정 → **투디지트**
- 역량: 합계 33, 금융 전공 + 기획 문서화 능력 보유
- 선호도: 2지망 투디지트, 1지망 디라브(난이도 최상으로 인해 조정)
- 기업 특성: 금융 서비스 구조 이해 + 데이터 분석 경험을 요구
- 결론: **금융 전공자로서 투자자 정보 서비스 기획에 적합**, 상위권 학생(김채민)과 함께 2인 팀 구성

---
### 5-3. 윤하영 → **대덕자산운용**
- 역량: 합계 38 (상위권), 데이터 분석/해석 모두 상
- 선호도: 1지망 디라브, 2지망 대덕자산운용
- 기업 난이도: 디라브는 기관 전략·이코노믹스 중심으로 난이도 최상 → 교육적 측면에서 조정
- 대덕자산운용은 금융상품 데이터를 다루는 서비스로 분석 역량과 잘 맞음
- 결론: **금융 데이터 기반 플랫폼 기획에 최적화된 배치**

---
### 5-4. 정예진 → **대덕자산운용**
- 역량: 합계 35, 문서화/시나리오/자료 정리 능력 강함
- 선호도: 1지망 디라브, 2지망 대덕자산운용
- 기업 특성: 서비스 구조 정의, 금융상품 정보 구조화, 문서화 작업 다수 필요
- 결론: **윤하영과 함께 2인 팀 구성 시, 분석 + 문서화 균형이 좋음**

---
### 5-5. 김유림 → **더빌딩**
- 역량: 합계 36, 텍스트 분석 및 엑셀 기반 데이터 정리에 강점
- 선호도: 1지망 더빌딩
- 기업 특성: 통화/문자/민원 로그 분석, 콜 유형 정의, 콜봇·사람 처리 영역 구분 → 텍스트 분석 능력이 직접적으로 활용됨
- 결론: **문의 유형 분류, 콜로그 분석 등에서 역량을 100% 활용할 수 있는 최적 매칭**

---
### 5-6. 신주협 → **씽크플랜두**
- 역량: 합계 30(중하위), 데이터 수집(DB 활용)은 강하지만 분석/해석은 중간 수준
- 선호도: 1지망 대덕자산운용, 2지망 코코넛사일로, 3지망 씽크플랜두
- 기업 난이도:
  - 대덕자산운용: 이미 상위권 학생 배정 + 분석 요구 수준 높음
  - 코코넛사일로: UX/IA 및 대규모 통합플랫폼 기획으로 난이도 상
- 씽크플랜두는 API 기반 서비스 흐름 기획 및 데이터 분석 보조 역할에 적합
- 결론: **API·결제 흐름을 이해하며 성장할 수 있는 현실적인 선택**

---
### 5-7. 오은상 → **헬로프렌즈**
- 역량: 정량 평가값 미기입(0) → 상대적으로 기초 수준으로 가정
- 선호도: 대덕자산운용, 투디지트, 씽크플랜두 등 데이터/금융 중심 기업 선호
- 다만, 상위권 학생들이 이미 해당 기업에 배정되고 난이도를 고려했을 때, 오은상에게는 **리서치·문서 중심 UX 보조 역할**이 더 적합
- 헬로프렌즈는 시장·사용자 리서치, 여정지도, 기획서·가이드 작성 등 리서치/문서 작업 비중이 커 입문자도 기여 가능
- 결론: **리서치 및 기획 문서 중심 역할로 경험을 쌓기 좋은 매칭**
