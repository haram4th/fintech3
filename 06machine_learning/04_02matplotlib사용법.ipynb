# 2X2 그래프에서 왼쪽 행 결합하기
* plt.subplot_mosaic([['left', 'right_top'],
                     ['left', 'right_bottom']])

```python
fig, ax = plt.subplot_mosaic([['left', 'right_top'],
                             ['left', 'right_bottom']])
plt.show()
```

# 2x2 그래프에서 오른쪽 행을 합치고 왼쪽은 2개

```python
fig, ax = plt.subplot_mosaic([['left_1', 'right'],
                             ['left_2', 'right']])
plt.show()
```

# matplotlib의 2가지 코딩 스타일

## 00-스타일: matplotlob의 공식 추천 스타일
* 복잡한 그래프를 그릴 때 사용
* 사용법이 pyplot 스타일보다 복잡함

```python
x = np.linspace(0, 2, 100)
x
```

```python
fig, ax = plt.subplots(figsize=(5, 2.7))
ax.plot(x, x, label='linear')
ax.plot(x, x**2, label='quadratic')
ax.plot(x, x**3, label='cubic')
ax.set_xlabel('x label')
ax.set_ylabel('y label')
ax.set_title('간단 그래프')
ax.legend()
plt.show()
```
