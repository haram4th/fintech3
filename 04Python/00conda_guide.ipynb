# Conda 사용법 교안

이 노트북은 Conda(가상환경/패키지 관리 도구)의 기본 개념부터 실습까지 포함합니다. 각 섹션을 순서대로 실행/참고하면서 학습하면 됩니다.

## 1. Conda란?

**Conda**는 파이썬을 포함한 다양한 언어용 패키지와 실행환경(가상환경)을 관리하는 도구입니다.

|기능|설명|
|---|---|
|패키지 관리|`numpy`, `pandas` 등 라이브러리를 설치/업데이트/삭제|
|환경 관리|프로젝트마다 독립된 파이썬 버전과 패키지 버전을 따로 유지|
|재현성|`environment.yml`로 같은 환경을 다른 컴퓨터에서 그대로 복제 가능|

➡ 데이터 분석/머신러닝 프로젝트에서는 **Conda 환경을 따로 만들어 쓰는 것**이 사실상 필수입니다.

## 2. Conda 기본 명령 구조

```bash
conda [명령어] [옵션]
```
예시:
```bash
conda install numpy
conda create -n myenv python=3.11
```

## 3. 가상환경(Environment) 관리

### 3.1 현재 conda 환경 목록 보기
```bash
conda env list
# 또는
conda info --envs
```

### 3.2 새로운 환경 만들기 (파이썬 버전까지 지정)
```bash
conda create -n fintech python=3.11
```
- `-n fintech` : 환경 이름이 `fintech`
- `python=3.11` : 이 환경은 Python 3.11로 시작

### 3.3 환경 활성화 / 비활성화
```bash
conda activate fintech
conda deactivate
```

### 3.4 환경 삭제
```bash
conda remove -n fintech --all
```

### 3.5 환경 복제 (완전한 카피)
```bash
conda create -n fintech_clone --clone fintech
```

## 4. 패키지 관리

### 4.1 패키지 설치
```bash
conda install pandas
```
특정 버전 지정:
```bash
conda install pandas=2.2.3
```

### 4.2 여러 패키지 한 번에 설치
```bash
conda install numpy pandas matplotlib
```

### 4.3 패키지 업데이트
```bash
conda update numpy
```

### 4.4 패키지 제거
```bash
conda remove numpy
```

### 4.5 현재 환경에 깔린 패키지 목록 보기
```bash
conda list
```

### 4.6 (중요) pip와 같이 쓰는 경우
1. 먼저 conda 환경을 활성화하고
2. 그 안에서 pip install 실행

```bash
conda activate fintech
pip install seaborn
```

👉 이렇게 하면 pip로 설치한 패키지도 같은 환경 안에 들어갑니다.

## 5. 환경 내보내기 & 불러오기 (복제/공유)

연구/협업/배포에서 아주 중요합니다. 나만 되는 환경 → 다른 컴퓨터에서도 그대로 재현.

### 5.1 현재 환경을 `.yml`로 저장하기
```bash
conda activate fintech
conda env export > environment.yml
```
이 파일에는
- 파이썬 버전
- conda 패키지들
- pip로 깐 패키지들까지
모두 기록됩니다.

### 5.2 `.yml`로 새 환경 만들기
```bash
conda env create -n fintech_clone -f environment.yml
```

### 5.3 기존 환경을 `.yml` 기준으로 업데이트(동기화)
```bash
conda activate fintech_clone
conda env update -f environment.yml --prune
```
`--prune`: yml에 없는 패키지는 제거해서 최대한 동일하게 맞춤.

## 6. 채널(channel) 관리

채널은 패키지를 어디서 가져올지 결정하는 소스 저장소입니다. 대표 채널은 `defaults`와 `conda-forge`.

### 6.1 채널 추가하기
```bash
conda config --add channels conda-forge
```

### 6.2 현재 채널 보기
```bash
conda config --show channels
```

일부 최신 패키지(예: 특정 버전의 `geopandas`, `cartopy`)는 `conda-forge` 쪽이 더 빠르게 제공되는 경우가 많습니다.

## 7. 유지보수 / 문제해결 명령들

|명령|설명|
|---|---|
|`conda clean --all`|캐시, 다운로드 찌꺼기 정리 (용량 확보)|
|`conda update conda`|conda 자체를 최신으로 업데이트|
|`conda update anaconda`|Anaconda 배포판 전체를 갱신|
|`conda info`|현재 conda의 상태, 경로, 활성 환경 등 정보 출력|

팁:
- base 환경은 가급적 건드리지 말고, 프로젝트마다 별도의 환경을 만드세요.
- 환경 이름은 짧고 의미 있게(`fintech`, `ml_class`, `webapp`).

## 9. 핵심 요약 (암기 포인트)

|기능|명령|
|---|---|
|환경 만들기|`conda create -n myenv python=3.11`|
|환경 활성화|`conda activate myenv`|
|환경 비활성화|`conda deactivate`|
|환경 삭제|`conda remove -n myenv --all`|
|패키지 설치|`conda install pandas`|
|패키지 제거|`conda remove pandas`|
|패키지 목록|`conda list`|
|환경 목록|`conda env list`|
|환경 내보내기|`conda env export > environment.yml`|
|환경 복원|`conda env create -f environment.yml`|

➡ 최소 이 표만 외워도 실무/수업 대부분 커버 가능합니다.
