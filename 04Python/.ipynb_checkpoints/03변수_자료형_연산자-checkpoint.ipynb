# 자료형(data type)
## 파이썬의 자료형은 동적으로 변화
* int: 정수(소숫점이 없는 숫자)
* float: 실수(소숫점이 있는 숫자)
* str: 문자 / 문자열
* list: 여러개의 자료가 순서대로 저장된 자료형, 모든 자료형을 넣어 만들 수 있음 [자료]
* tuple: list비슷하나 한 번 만들면 수정이 안됨, 모든 자료형을 넣어 만들 수 있음 (자료)
* dictionary(dict): key:value 쌍으로 이루어진 자료형, 모든 자료형 넣어 만들 수 있음 {key : value}
* set: 집합자료형, 자료의 중복이 되지 않음 {자료}

# 논리 연산자
* and, or, not

* and: 왼쪽과 오른쪽 값 모두 True 일 때 True, 하나라도 False면 False

```python
# True and True
3 == 3 and 5 > 3
```

```python
# False and True
3 < 3 and 5 > 3
```

```python
# True and False
3 == 3 and 5 < 3
```

```python
# False and False
3 < 3 and 5 < 3 
```

* or 왼쪽이나 오른쪽 중에 하나라도 True 면 True, 둘 다 False 일 때 만 False

```python
# True or True
3 == 3 or 5 > 3
```

```python
# False or True
3 > 3 or 5 > 3
```

```python
# True or False
3 == 3 or 5 < 3
```

```python
# False or False
3 > 3 or 5 < 3
```

* not 값을 반대로 뒤집음 True => False, False => True

```python
not True
```

```python
not False
```

```python
not 3 > 3 and not 5 < 3
```

* not, and, or의 순위 
* not > and > or

```python
False and False or True
not True and False or not False
```

```python
# True and True
10 == 10 and 10 != 5
```

```python
# == 는 값만 같으면 True 처리, is는 값이 같아도 데이터 타입다르면 False
print(a == b)
print(a is b)
```

```python
d = 3
```

```python
# a와 d 모두 int 3이기 때문에 True
print(a is d)
```

```python
# is not 은 둘이 다르면 True
print(a is not b)
```

```python
mem = [1,2,3,4,5, "육"]
```

```python
3 in mem
```
