# 반복문 for, while
* 반복문은 동일한 작업을 여러 번 반복해야 할 경우 사용
* for, while 두 종류의 반복문이 있음
* for는 **반복 횟수가 정해져 있는 경우** 사용
* while은 반복 횟수가 정해져 있지 않은 경우 사용 **(무한반복)**

* 1에서 50사이의 숫자중에서 짝수만 더한 합을 구하세요

  * 짝수인지 판별 if

## 반복문에 사용할 자료를 일부 구간만 사용하기
* 슬라이싱

* 1부터 100까지 숫자를 생성하고 그중에서 30-70까지만 더해서 출력하세요.

```python
# if 조건문을 이용해서 필터링
hap = 0
for num in range(1, 101):
#     print(num, end=" ")
    if 30 <= num <= 70:
        hap += num
        print(f"num: {num}, hap: {hap}")
# print(hap)
```

```python
# 슬라이싱 활용
hap = 0
for num in range(1, 101)[29:70]:
#     print(num, end=" ")
    hap += num
print(hap)
```

* 1-100까지 숫자중에서 짝수만 합 구하기 (슬라이싱 이용)

```python
hap = 0
for num in range(1, 101)[1::2]:
    print(num, end=" ")
    hap += num
print(hap)
```

* 1에서 100까지 숫자를 생성하고 역순으로 출력하세요

```python
# range 함수에서 숫자를 역순으로 만들기
for i in range(100, 0, -1):
    print(i, end=" ")
```

```python
# 슬라이싱을 이용한 역인덱싱
for i in range(1, 101)[::-1]:
    print(i, end=" ")
```

```python
# reversed() 함수 사용
for i in reversed(range(1, 101)):
    print(i, end=" ")
```

```python
# sorted(reverse=True) 함수 사용
for i in sorted(range(1, 101), reverse=True):
    print(i, end=" ")
```

# countinue / break
* countinue: 코드 실행을 건너 뛰고 반복 계속 진행
* break: 특정 조건이 되었을 때 반복문 정지

```python
for idx, i in enumerate(s):
    if idx % 2 == 0:
        continue
        print(idx, i)
    else:
        if idx > 19:
            break
            pass
        print(idx, i)
```

```python
# 1-100까지 숫자를 생성하고 홀수만 합계를 구하세요.
# 짝수면 짝수만 모인 리스트를 만들고 출력하세요.
# 숫자가 70을 초과하면 멈추세요.
hap = 0
even = []
for i in range(1, 101):
#     print(i, end=" ") 
    if i > 70:
        break
    else:
        if i % 2 != 0:
            hap += i
        else:
            even.append(i)
print(f"홀수의 합: {hap}")
print(f"짝수의 리스트: {even}")
        
```

# while: 반복할 횟수가 정해지지 않은 경우
* 무한 반복 가능
* 무한 반복이 되지 않도록 조건식을 잘 짜주는 것이 중요

변수 선언 및 초기화 num = 0<br>
while 조건식:<br>
____반복할 코드<br>
____변수 증가/감소 코드<br>
____반복문이 종료되는 조건 + break

* while 문으로 1-10까지 출력하기

```python
i = 1
while i <= 10:
    print(i, end =" ")
    i += 1
```

```python
i = 1
while True:
    print(i, end =" ")
    if i > 10:
        break
    i += 1
    
```

* 1-50까지 수 중에서 3의 배수이거나 짝수인 수를 찾아서 합계를 출력하세요. 

```python
i = 1
hap = 0
list_4s = []
while i <= 50:
    if i % 3 == 0 or i % 2 == 0:
        hap += i
        print(f"i: {i}, hap: {hap}")
    if i % 4 == 0:
        list_4s.append(i)
    i += 1
print(hap)
print(list_4s)
```

# 중첩 반복문
* 반복문을 2개 이상 겹쳐서 사용
* 시계의 원리와 같음
* 초 60 -> 1분 60분 -> 1시간 24시간
* 안쪽 반복문이 끝난 후 바깥쪽 반복문이 실행
* 아날로그 시계의 초침 -> 분침 -> 시침

* 구구단 만들기 2중 중첩 for문으로 작성
* 2단 - 9단
* 각 단마다 1-9
* 2 * 1 = 2
* 2 * 2 = 4

```python
# 2단 - 9단 시침
for num1 in range(2, 10):
    # 분침
    for num2 in range(1, 10):
        print(f"{num1} X {num2} = {num1 * num2}")
```

* 반복문 3개로 시계 만들기
* 초: 00-59
* 분: 00-59
* 시: 00-23
import time
time.sleep(0.1)
* 00:00:00 - 23:59:59

```python
import time
for si in range(0, 24):
    for bun in range(0, 60):
        for cho in range(0, 60):
            print(f"{si:02d}:{bun:02d}:{cho:02d}", end="\r")
            time.sleep(1)
```
